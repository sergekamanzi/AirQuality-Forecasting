```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import tensorflow as tf
import os
```

```python
# Load the datasets
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
```

#Data Preprocessing

```python
# Convert 'datetime' to datetime format
train['datetime'] = pd.to_datetime(train['datetime'])
test['datetime'] = pd.to_datetime(test['datetime'])
```

```python
# Set 'datetime' as index
train.set_index('datetime', inplace=True)
test.set_index('datetime', inplace=True)
```

```python
# Handle missing values by forward filling, then mean if needed
train.fillna(method='ffill', inplace=True)
train.fillna(train.mean(), inplace=True)
test.fillna(method='ffill', inplace=True)
test.fillna(test.mean(), inplace=True)
```

```python
# Separate features and target
X_train = train.drop(['pm2.5', 'No'], axis=1)
y_train = train['pm2.5']
X_test = test.drop(['No'], axis=1)
```

```python
# Normalize features
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Create sequences for LSTM (e.g., use 24 hours as sequence length)
def create_sequences(data, target, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i + seq_length])
        y.append(target[i + seq_length])
    return np.array(X), np.array(y)

seq_length = 24  # 24-hour sequences
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train.values, seq_length)
```

# Prepare test data sequences

```python
# For each test sample, use the last seq_length available data points
X_test_seq = []
test_indices = test.index[-len(X_test_scaled):]  # Store indices for submission
for i in range(len(X_test_scaled)):
    start_idx = max(0, i - seq_length + 1)
    sequence = X_test_scaled[start_idx:i + 1]
    if len(sequence) < seq_length:
        # Pad with zeros if sequence is shorter than seq_length
        padding = np.zeros((seq_length - len(sequence), X_test_scaled.shape[1]))
        sequence = np.vstack((padding, sequence))
    X_test_seq.append(sequence)
X_test_seq = np.array(X_test_seq)
```

#Model Design

```python
# Define LSTM model
model = Sequential([
    LSTM(64, activation='relu', input_shape=(seq_length, X_train.shape[1]), return_sequences=True),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
])
```

```python
# Compile the model with custom learning rate
optimizer = Adam(learning_rate=0.001)
model.compile(
    optimizer=optimizer,
    loss='mse',
    metrics=[tf.keras.metrics.RootMeanSquaredError()]
)
```

```python
# Display model architecture
model.summary()
```

#Training

```python
# Train the model
history = model.fit(
    X_train_seq, y_train_seq,
    epochs=30,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)
```

#Results and Visualization

```python
# Plot training and validation loss
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss (MSE)')
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()
```

```python
# Calculate final training RMSE
train_predictions = model.predict(X_train_seq)
train_rmse = np.sqrt(np.mean((y_train_seq - train_predictions.flatten())**2))
print(f"Final Training RMSE: {train_rmse}")
```

#Predictions

```python
# Make predictions on test set
predictions = model.predict(X_test_seq)

# Ensure predictions are non-negative and handle NaN
predictions = np.nan_to_num(predictions, nan=0.0)
predictions = np.maximum(predictions, 0)

# Convert predictions to integers
predictions = np.round(predictions).astype(int)

# Prepare submission file
submission = pd.DataFrame({
    'row ID': pd.to_datetime(test_indices).strftime('%Y-%m-%d %-H:%M:%S'),
    'pm2.5': predictions.flatten()
})

# Sort by 'row ID'
submission = submission.sort_values(by='row ID')

# Save submission in the same directory as input files
output_path = '/content/submission.csv'
submission.to_csv(output_path, index=False)
print(f"Submission file saved as '{output_path}'")
```

```python
# Import necessary libraries 5706.2665
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import tensorflow as tf
from sklearn.feature_selection import SelectKBest, f_regression
import seaborn as sns

# Set random seed for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Load the datasets
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')

# Data Exploration
print("Training Data Overview:")
print(train.head())
print("\nColumn Names:")
print(train.columns)
print("\nMissing Values in Train:")
print(train.isnull().sum())
print("\nMissing Values in Test:")
print(test.isnull().sum())
print("\nTraining Data Summary:")
print(train.describe())

# Convert 'datetime' to datetime format
train['datetime'] = pd.to_datetime(train['datetime'])
test['datetime'] = pd.to_datetime(test['datetime'])

# Set 'datetime' as index
train.set_index('datetime', inplace=True)
test.set_index('datetime', inplace=True)

# Improved Missing Value Handling
# Interpolate missing pm2.5 values to preserve temporal patterns
train['pm2.5'] = train['pm2.5'].interpolate(method='time')
test = test.interpolate(method='time')  # Interpolate test set features
# Fill any remaining missing values with mean
train.fillna(train.mean(), inplace=True)
test.fillna(test.mean(), inplace=True)

# Feature Engineering
# Add time-based features
train['hour'] = train.index.hour
train['day'] = train.index.day
train['month'] = train.index.month
test['hour'] = test.index.hour
test['day'] = test.index.day
test['month'] = test.index.month

# Feature Selection
# Drop 'No' as it’s an identifier
X_train = train.drop(['pm2.5', 'No'], axis=1)
y_train = train['pm2.5']
X_test = test.drop(['No'], axis=1)

# Select top 6 features based on correlation with pm2.5
selector = SelectKBest(score_func=f_regression, k=6)
selector.fit(X_train, y_train)
selected_features = X_train.columns[selector.get_support()]
print("\nSelected Features:", selected_features)
X_train = X_train[selected_features]
X_test = X_test[selected_features]

# Normalize features
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Create sequences for LSTM
def create_sequences(data, target, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i + seq_length])
        y.append(target[i + seq_length])
    return np.array(X), np.array(y)

seq_length = 24  # 24-hour sequences
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train.values, seq_length)

# Prepare test data sequences with improved padding
X_test_seq = []
test_indices = test.index[-len(X_test_scaled):]
for i in range(len(X_test_scaled)):
    start_idx = max(0, i - seq_length + 1)
    sequence = X_test_scaled[start_idx:i + 1]
    if len(sequence) < seq_length:
        # Pad with the first available value instead of zeros
        padding = np.repeat(sequence[0:1], seq_length - len(sequence), axis=0)
        sequence = np.vstack((padding, sequence))
    X_test_seq.append(sequence)
X_test_seq = np.array(X_test_seq)

# Define Improved LSTM Model
model = Sequential([
    Bidirectional(LSTM(128, activation='tanh', return_sequences=True), input_shape=(seq_length, X_train.shape[1])),
    Dropout(0.3),
    LSTM(64, activation='tanh'),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1)
])

# Compile the model
optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='mse', metrics=[tf.keras.metrics.RootMeanSquaredError()])

# Callbacks for early stopping and learning rate reduction
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
lr_scheduler = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)

# Display model architecture
model.summary()

# Train the model
history = model.fit(
    X_train_seq, y_train_seq,
    epochs=200,  # Increased epochs, relying on early stopping
    batch_size=16,  # Smaller batch size for better generalization
    validation_split=0.2,
    callbacks=[early_stopping, lr_scheduler],
    verbose=1
)

# Plot training and validation loss
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss (MSE)')
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()

# Calculate final training RMSE
train_predictions = model.predict(X_train_seq)
train_rmse = np.sqrt(np.mean((y_train_seq - train_predictions.flatten())**2))
print(f"Final Training RMSE: {train_rmse}")

# Make predictions on test set
predictions = model.predict(X_test_seq)
predictions = np.nan_to_num(predictions, nan=0.0)
predictions = np.maximum(predictions, 0)  # Ensure non-negative predictions

# Prepare submission file
submission = pd.DataFrame({
    'row ID': pd.to_datetime(test_indices).strftime('%Y-%m-%d %-H:%M:%S'),
    'pm2.5': predictions.flatten()  # No rounding to preserve precision
})
submission = submission.sort_values(by='row ID')
output_path = '/content/submission.csv'
submission.to_csv(output_path, index=False)
print(f"Submission file saved as '{output_path}'")
```
