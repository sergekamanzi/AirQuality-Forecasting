```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow as tf
```

```python
# Load the datasets
# Ensure train.csv and test.csv are in your Google Drive folder
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')
```

```python
%%capture
# --- Data Exploration ---
# Inspect the dataset to understand its structure and identify patterns
print("Training Data Overview:")
print(train.head())
print("\nTraining Data Summary:")
print(train.describe())
print("\nMissing Values:")
print(train.isnull().sum())
```

```python
# Convert 'datetime' column to datetime format and set as index
train['datetime'] = pd.to_datetime(train['datetime'])
test['datetime'] = pd.to_datetime(test['datetime'])
train.set_index('datetime', inplace=True)
test.set_index('datetime', inplace=True)
```

```python
# Visualize PM2.5 time series
plt.figure(figsize=(12, 6))
plt.plot(train.index, train['pm2.5'], label='PM2.5')
plt.title('PM2.5 Concentration Over Time')
plt.xlabel('Date')
plt.ylabel('PM2.5')
plt.legend()
plt.show()

# Correlation heatmap to identify feature relationships
plt.figure(figsize=(10, 8))
sns.heatmap(train.corr(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
# --- Data Preprocessing ---
# Handle missing values using linear interpolation for time-series data
train.interpolate(method='linear', inplace=True)
test.interpolate(method='linear', inplace=True)

# Separate features and target
X_train = train.drop(['pm2.5', 'No'], axis=1)
y_train = train['pm2.5']
X_test = test.drop(['No'], axis=1)

# Scale features and target using MinMaxScaler
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
y_scaler = MinMaxScaler()
y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1, 1))
```

```python
# Create sequences for LSTM input (e.g., 24 timesteps)
def create_sequences(X, y, time_steps=24):
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X[i:(i + time_steps)])
        ys.append(y[i + time_steps])
    return np.array(Xs), np.array(ys)

time_steps = 24
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, time_steps)
```

```python
# Split data into training and validation sets
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_seq, y_train_seq, test_size=0.2, shuffle=False
)
```

```python
# --- Define and Train LSTM Model ---
# Build a robust LSTM model with dropout to prevent overfitting
model = Sequential([
    LSTM(64, activation='relu', input_shape=(time_steps, X_train_scaled.shape[1]), return_sequences=True),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
])

# Compile the model with RMSE metric
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='mse',
    metrics=[tf.keras.metrics.RootMeanSquaredError(name='rmse')]
)
model.summary()

# Train the model with early stopping
early_stopping = EarlyStopping(monitor='val_rmse', patience=5, restore_best_weights=True)
history = model.fit(
    X_train_split, y_train_split,
    validation_data=(X_val_split, y_val_split),
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping]
)
```

```python
# --- Visualize Training Performance ---
# Plot training and validation loss/RMSE
plt.figure(figsize=(12, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.plot(history.history['rmse'], label='Training RMSE')
plt.plot(history.history['val_rmse'], label='Validation RMSE')
plt.title('Training and Validation Loss/RMSE')
plt.xlabel('Epochs')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
# --- Conduct Experiments ---
# Function to run experiments with different hyperparameters
def run_experiment(units, layers, lr, time_steps, dropout_rate):
    X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, time_steps)
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
        X_train_seq, y_train_seq, test_size=0.2, shuffle=False
    )
    model = Sequential()
    model.add(LSTM(units, activation='relu', input_shape=(time_steps, X_train_scaled.shape[1]), return_sequences=(layers > 1)))
    if layers > 1:
        model.add(LSTM(units//2, activation='relu'))
    model.add(Dropout(dropout_rate))
    model.add(Dense(1))
    model.compile(optimizer=Adam(learning_rate=lr), loss='mse', metrics=['rmse'])
    history = model.fit(
        X_train_split, y_train_split, validation_data=(X_val_split, y_val_split),
        epochs=50, batch_size=32, callbacks=[early_stopping], verbose=0
    )
    return min(history.history['val_rmse'])

# Run experiments with different configurations
experiments = [
    {'units': 32, 'layers': 1, 'lr': 0.001, 'time_steps': 24, 'dropout': 0.2},
    {'units': 64, 'layers': 2, 'lr': 0.0005, 'time_steps': 24, 'dropout': 0.3},
    {'units': 128, 'layers': 2, 'lr': 0.001, 'time_steps': 12, 'dropout': 0.2}
]
results = []
for exp in experiments:
    rmse = run_experiment(**exp)
    results.append({**exp, 'val_rmse': rmse})

# Create and display experiment table
results_df = pd.DataFrame(results)
print("Experiment Results:")
print(results_df)

# --- Generate Predictions for Test Data ---
# Create sequences for test data
X_test_seq = []
for i in range(len(X_test_scaled) - time_steps):
    X_test_seq.append(X_test_scaled[i:(i + time_steps)])
X_test_seq = np.array(X_test_seq)

# Predict and inverse-scale predictions
predictions = model.predict(X_test_seq)
predictions = y_scaler.inverse_transform(predictions)

# Prepare submission file
submission = pd.DataFrame({
    'row ID': pd.to_datetime(test.index[time_steps:]).strftime('%Y-%m-%d %H:%M:%S'),
    'pm2.5': predictions.flatten()
})
submission = submission.sort_values(by='row ID')
submission.to_csv('/content/drive/MyDrive/Kaggle_competition_ML/air_quality_forcasting/submission.csv', index=False)

# Display submission file preview
print("Submission File Preview:")
print(submission.head())
```

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow as tf

# Load the datasets
train = pd.read_csv('/content/train.csv')
test = pd.read_csv('/content/test.csv')

# --- Data Exploration ---
print("Training Data Overview:")
print(train.head())
print("\nTraining Data Summary:")
print(train.describe())
print("\nMissing Values:")
print(train.isnull().sum())
print(f"Test dataset size: {len(test)}")  # Verify test dataset size

# Convert 'datetime' column to datetime format and set as index
train['datetime'] = pd.to_datetime(train['datetime'])
test['datetime'] = pd.to_datetime(test['datetime'])
train.set_index('datetime', inplace=True)
test.set_index('datetime', inplace=True)

# Ensure test index is sorted
test = test.sort_index()

# Visualize PM2.5 time series
plt.figure(figsize=(12, 6))
plt.plot(train.index, train['pm2.5'], label='PM2.5')
plt.title('PM2.5 Concentration Over Time')
plt.xlabel('Date')
plt.ylabel('PM2.5')
plt.legend()
plt.show()

# Correlation heatmap to identify feature relationships
plt.figure(figsize=(10, 8))
sns.heatmap(train.corr(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()

# --- Data Preprocessing ---
train.interpolate(method='linear', inplace=True)
test.interpolate(method='linear', inplace=True)

X_train = train.drop(['pm2.5', 'No'], axis=1)
y_train = train['pm2.5']
X_test = test.drop(['No'], axis=1)

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
y_scaler = MinMaxScaler()
y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1, 1))

def create_sequences(X, y=None, time_steps=24, is_test=False):
    Xs = []
    if not is_test:
        ys = []
        for i in range(len(X) - time_steps):
            Xs.append(X[i:(i + time_steps)])
            ys.append(y[i + time_steps])
        return np.array(Xs), np.array(ys)
    else:
        for i in range(len(X) - time_steps + 1):
            Xs.append(X[i:(i + time_steps)])
        return np.array(Xs)

time_steps = 24
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, time_steps, is_test=False)

X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_seq, y_train_seq, test_size=0.2, shuffle=False
)

# --- Define and Train LSTM Model ---
model = Sequential([
    LSTM(64, activation='relu', input_shape=(time_steps, X_train_scaled.shape[1]), return_sequences=True),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(optimizer=Adam(learning_rate=0.001), loss='mse', metrics=[tf.keras.metrics.RootMeanSquaredError()])
model.summary()

early_stopping = EarlyStopping(monitor='val_root_mean_squared_error', patience=5, restore_best_weights=True)
history = model.fit(X_train_split, y_train_split, validation_data=(X_val_split, y_val_split), epochs=50, batch_size=32, callbacks=[early_stopping])

# --- Visualize Training Performance ---
plt.figure(figsize=(12, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.plot(history.history['root_mean_squared_error'], label='Training RMSE')
plt.plot(history.history['val_root_mean_squared_error'], label='Validation RMSE')
plt.title('Training and Validation Loss/RMSE')
plt.xlabel('Epochs')
plt.ylabel('Value')
plt.legend()
plt.show()

# --- Conduct Experiments ---
def run_experiment(units, layers, lr, time_steps, dropout_rate):
    X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, time_steps, is_test=False)
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(X_train_seq, y_train_seq, test_size=0.2, shuffle=False)
    model = Sequential()
    model.add(LSTM(units, activation='relu', input_shape=(time_steps, X_train_scaled.shape[1]), return_sequences=(layers > 1)))
    if layers > 1:
        model.add(Dropout(dropout_rate))
        model.add(LSTM(units//2, activation='relu'))
    model.add(Dropout(dropout_rate))
    model.add(Dense(1))
    model.compile(optimizer=Adam(learning_rate=lr), loss='mse', metrics=[tf.keras.metrics.RootMeanSquaredError()])
    history = model.fit(X_train_split, y_train_split, validation_data=(X_val_split, y_val_split), epochs=50, batch_size=32, callbacks=[EarlyStopping(monitor='val_root_mean_squared_error', patience=5, restore_best_weights=True)], verbose=0)
    return min(history.history['val_root_mean_squared_error'])

experiments = [
    {'units': 32, 'layers': 1, 'lr': 0.001, 'time_steps': 24, 'dropout_rate': 0.2},
    {'units': 64, 'layers': 2, 'lr': 0.0005, 'time_steps': 24, 'dropout_rate': 0.3},
    {'units': 128, 'layers': 2, 'lr': 0.001, 'time_steps': 12, 'dropout_rate': 0.2}
]
results = []
for exp in experiments:
    rmse = run_experiment(**exp)
    results.append({**exp, 'val_rmse': rmse})

results_df = pd.DataFrame(results)
print("Experiment Results:")
print(results_df)

# --- Generate Predictions for Test Data ---
# Step 1: Initial predictions for available sequences
X_test_seq = create_sequences(X_test_scaled, time_steps=time_steps, is_test=True)
initial_predictions = model.predict(X_test_seq)
initial_predictions = y_scaler.inverse_transform(initial_predictions)

# Step 2: Iteratively predict the remaining rows to reach 13,148
expected_rows = 13148
num_initial_predictions = len(initial_predictions)
print(f"Number of initial predictions: {num_initial_predictions}")

if num_initial_predictions >= expected_rows:
    predictions = initial_predictions[:expected_rows]
else:
    remaining_rows = expected_rows - num_initial_predictions
    all_predictions = list(initial_predictions.flatten())
    current_sequence = X_test_seq[-1].copy()  # Start with the last sequence

    for _ in range(remaining_rows):
        next_pred = model.predict(current_sequence.reshape(1, time_steps, current_sequence.shape[1]), verbose=0)
        next_pred = y_scaler.inverse_transform(next_pred)[0, 0]
        all_predictions.append(next_pred)
        current_sequence = np.roll(current_sequence, -1, axis=0)
        current_sequence[-1] = current_sequence[-2]  # Use last feature values as a proxy

    predictions = np.array(all_predictions)

# Verify the number of predictions
print(f"Total number of predictions: {len(predictions)}")
if len(predictions) != expected_rows:
    raise ValueError(f"Expected {expected_rows} predictions, but got {len(predictions)}.")

# Step 3: Handle submission_index to match 13,148 rows
# Calculate the required length for submission_index
required_test_length = time_steps - 1 + expected_rows  # 23 + 13148 = 13171
if len(test) < required_test_length:
    print(f"Warning: Test dataset has {len(test)} rows, but {required_test_length} are needed. Extrapolating datetime index.")
    # Calculate the time difference between consecutive timestamps (assume hourly data)
    time_diff = test.index[1] - test.index[0] if len(test) > 1 else pd.Timedelta(hours=1)
    # Extend the index by adding timestamps
    last_timestamp = test.index[-1]
    additional_timestamps = [last_timestamp + (i + 1) * time_diff for i in range(required_test_length - len(test))]
    extended_index = pd.DatetimeIndex(list(test.index) + additional_timestamps)
else:
    extended_index = test.index

# Use the first 13,148 timestamps starting after the time_steps offset
submission_index = extended_index[time_steps-1:time_steps-1+expected_rows]

# Verify lengths match
print(f"Length of submission_index: {len(submission_index)}")
if len(submission_index) != expected_rows:
    raise ValueError(f"Expected {expected_rows} timestamps in submission_index, but got {len(submission_index)}.")

# Prepare submission file
submission = pd.DataFrame({
    'row ID': pd.to_datetime(submission_index).strftime('%Y-%m-%d %H:%M:%S'),
    'pm2.5': predictions
})

# Verify submission size and check for null values
print(f"Submission size: {len(submission)} rows")
if submission['pm2.5'].isnull().any():
    raise ValueError("Submission contains null values, which is not allowed.")
if len(submission) != expected_rows:
    raise ValueError(f"Submission has {len(submission)} rows, expected {expected_rows}.")

submission = submission.sort_values(by='row ID')
submission.to_csv('/content/submission.csv', index=False)

# Display submission file preview
print("Submission File Preview:")
print(submission.head())
```
